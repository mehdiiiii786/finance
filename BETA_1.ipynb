```python
import pandas as pd
import numpy as np
import yfinance as yf
from sklearn.linear_model import LinearRegression
import plotly.express as px
```

```python
class MarketAnalyzer :
    def __init__(self, ticker) :
        self.ticker = ticker
        self.ticker_machine = yf.Ticker(self.ticker)
        sdata = self.ticker_machine.history(period = "1y")
        mdata = yf.Ticker("^GSPC").history(period = "1y")
        sreturn = sdata['Close'].pct_change().dropna()
        mreturn = mdata['Close'].pct_change().dropna()
        self.data = pd.DataFrame({'Stock returns' : sreturn,'Market returns' : mreturn}).dropna()
        print(self.data)
    def train_regression(self) :
      X = self.data[['Market returns']]
      Y = self.data['Stock returns']
      self. model = LinearRegression()
      self.model.fit(X,Y)
      self.beta = self.model.coef_[0]
      self.alpha = self.model.intercept_

    def plot_beta(self):
      X = self.data[['Market returns']]
      Y = self.data['Stock returns']
      r2 = self.model.score(X,Y)
      print(f"Model Confidence (R-Squared): {r2 * 100:.2f}%")
      print(f"This means {r2 * 100:.2f}% of {self.ticker}'s movement is explained by the S&P 500.")
      fig = px.scatter( self.data, x = 'Market returns', y = 'Stock returns', title = f"CAPM Beta ENGINE  : {self.ticker}  vs S&P500", trendline="ols",
      trendline_color_override="red")
      fig.show()


```

```python
LMD = MarketAnalyzer("LMT")
```

```python
T = LMD.train_regression()
T
```

```python
P = LMD.plot_beta()
```
